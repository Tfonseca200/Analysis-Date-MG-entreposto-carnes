## Regressão Linear: Outliers Inferiores - OSRM vs Distância Real

```python
# Calcular a diferença entre distância real e OSRM
df['distance_diff'] = df['actual_distance_to_destination'] - df['osrm_distance']

# Calcular IQR para identificar outliers inferiores
Q1 = df['distance_diff'].quantile(0.25)
Q3 = df['distance_diff'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR

# Filtrar apenas os outliers inferiores
lower_outliers = df[df['distance_diff'] < lower_bound]

# Regressão linear para os outliers inferiores
outlier_df = lower_outliers[['osrm_distance', 'actual_distance_to_destination']].dropna()
X = outlier_df['osrm_distance']
y = outlier_df['actual_distance_to_destination']
X_const = sm.add_constant(X)
model = sm.OLS(y, X_const).fit()

# Plotar pontos e linha da regressão linear
plt.figure(figsize=(10, 6))
plt.scatter(X, y, alpha=0.5, label='Outliers Inferiores')
plt.plot(X, model.fittedvalues, color='red', label='Regressão Linear (Outliers)')
plt.title('Regressão Linear: Outliers Inferiores - OSRM vs Distância Real')
plt.xlabel('Distância OSRM (km)')
plt.ylabel('Distância Real (km)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Agrupa os outliers inferiores pelas combinações de origem e destino,
# contando quantas vezes cada par (rota) aparece como outlier
frequent_issues = lower_outliers.groupby(['source_name', 'destination_name']).size().reset_index(name='count')

# Ordena as rotas pela contagem de outliers, da maior para a menor,
# e exibe as 10 rotas mais problemáticas
frequent_issues.sort_values(by='count', ascending=False).head(10)
```

**"Atraso na Entrega X Condições de Tráfego"**


"Aqui, dividimos o dia todo em 4 partes:
High\_traffic\_1: 8h às 12h
Medium\_Traffic: 12h às 17h
High\_Traffic\_2: 17h às 22h
Low\_Traffic: 22h às 8h

Estamos tentando descobrir se existe alguma relação entre o atraso nas entregas e o horário do dia em que essas viagens começaram."


```python
# Passo 1: Atribuir categorias de tráfego.
def assign_traffic_category(hour):
    if 8 <= hour < 12:
        return 'High Traffic_1'
    elif 12 <= hour < 17:
        return 'Medium Traffic'
    elif 17 <= hour < 22:
        return 'High Traffic_2'
    else:
        return 'Low Traffic'

# Extrair a hora de od_start_time e atribuir uma categoria de tráfego.
df['traffic_category'] = df['od_start_time'].dt.hour.map(assign_traffic_category)

# Passo 2: Agrupar por rota e situação do tráfego e contar a frequência.
traffic_frequency = (
    df.groupby(['source_name', 'destination_name', 'traffic_category'])
    .size()
    .reset_index(name='frequency')
)

# Passo 3: Reordenar as categorias de tráfego e ordenar por frequência.
traffic_order = ['High Traffic_1', 'High Traffic_2', 'Medium Traffic', 'Low Traffic']
traffic_frequency['traffic_category'] = pd.Categorical(
    traffic_frequency['traffic_category'],
    categories=traffic_order,
    ordered=True
)

# Ordenar por categoria de tráfego e frequência.
sorted_traffic_frequency = traffic_frequency.sort_values(
    by=['traffic_category', 'frequency'], ascending=[True, False]
)

# Passo 4: Calcular a porcentagem de remessas para cada categoria de tráfego.
total_shipments = sorted_traffic_frequency['frequency'].sum()
sorted_traffic_frequency['percentage_of_shipments'] = (sorted_traffic_frequency['frequency'] / total_shipments) * 100

# Passo 5: Exibir a tabela ordenada com porcentagem.
from IPython.display import display
display(sorted_traffic_frequency)

# Passo 6: Agrupar por categoria de tráfego e contar a frequência das condições gerais do trânsito.
traffic_condition_frequency = (
    df.groupby('traffic_category')
    .size()
    .reset_index(name='frequency')
)

# Passo 7: Calcular a porcentagem de remessas para as condições gerais do trânsito.
traffic_condition_frequency['percentage_of_shipments'] = (traffic_condition_frequency['frequency'] / total_shipments) * 100

# Passo 8: Ordenar por frequência em ordem decrescente.
traffic_condition_frequency_sorted = traffic_condition_frequency.sort_values(by='frequency', ascending=False)

# Passo 9: Exibir a tabela de condições de tráfego ordenada.
display(traffic_condition_frequency_sorted)
```

Esse código realiza uma análise completa do tráfego com base no horário de início das viagens, usando o campo od_start_time. Ele classifica as viagens conforme a hora do dia, conta a frequência por rota e calcula porcentagens.

"A tabela acima mostra que 28% do total de remessas atrasadas realmente começaram durante condições de tráfego intenso, o que é significativo.

Teste de Hipóteses:

Hipótese Nula (H₀): Os veículos se atrasaram devido às viagens que começaram em condições de Tráfego Intenso (tanto High Traffic\_1 quanto High Traffic\_2).
Hipótese Alternativa (H₁): Não há uma relação significativa entre os atrasos nas viagens e as condições de Tráfego Intenso."


Criando a tabela de resultados para realizar o teste de hipóteses"


```python
# Passo 1: Filtrar as linhas onde o Tempo Real é maior que o Tempo OSRM.
time_delays = df[df['time_diff_minutes'] > 0]

# Passo 2: Adicionar as colunas necessárias.
result_table = time_delays[['source_name', 'destination_name', 'od_start_time', 'traffic_category', 'time_diff_minutes']]

# Passo 3: Renomear colunas para maior clareza.
result_table = result_table.rename(columns={
    'od_start_time': 'trip_creation_time',
    'time_diff_minutes': 'Actual_minus_OSRM_time'
})

# Passo 4: Exibir a nova tabela.
from IPython.display import display
display(result_table)
```

Horário de início da viagem

Categoria de tráfego no momento da partida

Quanto tempo a viagem demorou além do esperado (em minutos)



```python
# Passo 1: Combinar "High Traffic_1" e "High Traffic_2" em uma única categoria chamada "High Traffic".
result_table['traffic_condition_combined'] = result_table['traffic_category'].replace({
    'High Traffic_1': 'High Traffic',
    'High Traffic_2': 'High Traffic'
})

# Filtrar dados para viagens onde Tempo Real > Tempo OSRM (viagens atrasadas).
delayed_trips_result = result_table[result_table['Actual_minus_OSRM_time'] > 0]

# Passo 2: Dividir os dados em dois grupos: High Traffic e Other Traffic
high_traffic_group = delayed_trips_result[delayed_trips_result['traffic_condition_combined'] == 'High Traffic']['Actual_minus_OSRM_time']
other_traffic_group = delayed_trips_result[delayed_trips_result['traffic_condition_combined'] != 'High Traffic']['Actual_minus_OSRM_time']
from scipy import stats
# Passo 3: Realizar o teste t.
t_stat, p_value = stats.ttest_ind(high_traffic_group, other_traffic_group, nan_policy='omit')

# Passo 4: Imprimir os resultados.
print(f"Estatística t: {t_stat}")
print(f"Valor-p: {p_value}")

# Interpretação.
if p_value < 0.05:
    print("Falha em Rejeitar a Hipótese Nula: Existe uma relação significativa entre os atrasos nas viagens e as condições de Tráfego Intenso.")
else:
    print("Rejeitar a Hipótese Nula: Os veículos não se atrasaram significativamente devido às viagens que começaram em condições de Tráfego Intenso.")
```

Se o valor-p for menor que 0.05, significa que há evidência estatística de que as viagens iniciadas em horários de tráfego intenso têm atrasos significativamente maiores.

Se o valor-p for maior que 0.05, indica que não há evidência suficiente para afirmar que o tráfego intenso causa atrasos maiores nas viagens.

```python
# Aplicar agrupamento KMeans.
# Inicializa o algoritmo KMeans para agrupar os dados em 3 clusters (grupos), com uma seed fixa para reprodutibilidade.
kmeans = KMeans(n_clusters=3, random_state=0).fit(cleaned_coords)

# Obtém os rótulos (labels) de cluster atribuídos a cada ponto de coordenada pelo KMeans.
labels = kmeans.labels_

# Como as coordenadas foram concatenadas (origem + destino), dividimos os rótulos em duas partes:
# A primeira parte corresponde aos clusters das coordenadas de origem.
source_labels = labels[:len(grouped_summary)]

# A segunda parte corresponde aos clusters das coordenadas de destino.
destination_labels = labels[len(grouped_summary):]

# Adiciona esses rótulos de cluster ao DataFrame original para cada viagem,
# permitindo identificar a qual grupo (cluster) cada estado de origem e destino pertence.
grouped_summary['source_cluster'] = source_labels
grouped_summary['destination_cluster'] = destination_labels
```

# Eficiência da rota por origem e horários de pico.

### Criei uma função chamada is_peak para definir os horários de pico.

### Primeiro, analisamos as 10 principais rotas (nome da origem) onde a diferença de tempo entre segment_actual_time e segment_osrm_time é maior.

```python
# Converter a coluna 'start_time' para o formato datetime para facilitar a manipulação de horários.
grouped_summary['start_time'] = pd.to_datetime(grouped_summary['start_time'])

# Definir uma função que verifica se um horário está dentro dos períodos de pico (8h-10h e 17h-19h).
def is_peak_hour(timestamp):
    hour = timestamp.hour
    return (8 <= hour < 10) or (17 <= hour < 19)

# Aplicar a função para criar uma nova coluna booleana 'is_peak' indicando se o horário da viagem está em pico.
grouped_summary['is_peak'] = grouped_summary['start_time'].apply(is_peak_hour)

# Filtrar os dados para considerar somente as viagens que ocorreram nos horários de pico.
peak_hours_data = grouped_summary[grouped_summary['is_peak']]

# Agrupar por rotas (origem e destino) e calcular a média do tempo real por segmento e do tempo previsto pelo OSRM.
route_efficiency_peak = peak_hours_data.groupby(['source_name', 'destination_name']).agg({
    'segment_actual_time': 'mean',
    'segment_osrm_time': 'mean'
}).reset_index()

# Calcular a diferença média entre o tempo real e o tempo previsto pelo OSRM para cada rota durante os horários de pico.
route_efficiency_peak['time_difference'] = route_efficiency_peak['segment_actual_time'] - route_efficiency_peak['segment_osrm_time']

# Selecionar as 10 rotas com a maior diferença média de tempo, ou seja, as rotas mais atrasadas nos horários de pico.
top_routes = route_efficiency_peak.nlargest(10, 'time_difference')
```

```python
# Cria uma figura grande para abrigar dois gráficos lado a lado.
plt.figure(figsize=(12, 6))

# Primeiro gráfico (à esquerda): gráfico de barras mostrando o tempo médio real das 10 principais rotas durante os horários de pico.
plt.subplot(1, 2, 1)
sns.barplot(data=top_routes, x='source_name', y='segment_actual_time', palette='viridis')
plt.title('Top 10 Rotas: Tempo Médio Real Durante Horários de Pico')
plt.xlabel('Rota')
plt.ylabel('Tempo Médio Real (minutos)')
plt.xticks(rotation=45, ha='right')  # Rotaciona os nomes das rotas para facilitar a leitura.

# Segundo gráfico (à direita): gráfico de barras mostrando a diferença de tempo (atraso médio) nas mesmas rotas durante horários de pico.
plt.subplot(1, 2, 2)
sns.barplot(data=top_routes, x='source_name', y='time_difference', palette='viridis')
plt.title('Top 10 Rotas: Diferença de Tempo Durante Horários de Pico')
plt.xlabel('Rota')
plt.ylabel('Diferença de Tempo (minutos)')
plt.xticks(rotation=45, ha='right')  # Rotaciona os nomes das rotas para facilitar a leitura.

# Ajusta o layout para que os gráficos não se sobreponham.
plt.tight_layout()

# Exibe os dois gráficos.
plt.show()
```

Este código gera dois gráficos lado a lado para as 10 rotas com maiores atrasos nos horários de pico. O primeiro mostra o tempo médio real que as viagens levaram nessas rotas, e o segundo mostra o atraso médio em relação ao tempo previsto (OSRM). Isso ajuda a visualizar quais rotas são mais problemáticas em termos de tempo durante os períodos críticos do dia.

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Dados da regressão
x = top_routes['segment_actual_time'].values.reshape(-1, 1)
y = top_routes['time_difference'].values

# Ajusta modelo
model = LinearRegression()
model.fit(x, y)
a = model.coef_[0]
b = model.intercept_
y_pred = model.predict(x)

# Gráfico
plt.figure(figsize=(10, 6))
sns.scatterplot(x=top_routes['segment_actual_time'], y=top_routes['time_difference'], s=80)
plt.plot(top_routes['segment_actual_time'], y_pred, color='red', linestyle='--',
         label=f'Regressão linear: y = {a:.2f}')

plt.title('Relação entre Tempo Médio Real e Atraso nas Top 10 Rotas')
plt.xlabel('Tempo Médio Real (minutos)')
plt.ylabel('Diferença de Tempo (atraso, minutos)')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Mostra equação abaixo do gráfico
plt.figtext(0.5, -0.06, f'Equação da regressão linear: y = {a:.2f}',
            wrap=True, horizontalalignment='center', fontsize=10)

plt.show()
```

# Rotas atípicas.

### Por último, mas não menos importante, tentamos identificar algumas rotas atípicas.

```python
# Calcular tempos médios por rota.
# Agrupa os dados por 'trip_uuid' e calcula a média do tempo real e do tempo estimado (OSRM) por segmento.
# Também mantém o primeiro nome de origem e destino para identificar a rota.
outlier_summary = grouped_summary.groupby('trip_uuid').agg({
    'segment_actual_time': 'mean',
    'segment_osrm_time': 'mean',
    'source_name': 'first',
    'destination_name': 'first'
}).reset_index()

# Identificar valores atípicos usando o método IQR.
# Calcula o primeiro quartil (Q1) e o terceiro quartil (Q3) do tempo real.
Q1 = outlier_summary['segment_actual_time'].quantile(0.25)
Q3 = outlier_summary['segment_actual_time'].quantile(0.75)
# Calcula o intervalo interquartil (IQR).
IQR = Q3 - Q1

# Definir limites para valores atípicos.
# Define os limites inferior e superior para detectar valores fora da faixa normal.
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identificar rotas atípicas.
# Cria uma nova coluna booleana que indica se o tempo da rota está fora dos limites definidos.
outlier_summary['is_outlier'] = outlier_summary['segment_actual_time'].apply(
    lambda x: x < lower_bound or x > upper_bound)

# Extrair rotas atípicas.
# Filtra apenas as rotas consideradas outliers.
outlier_routes = outlier_summary[outlier_summary['is_outlier']]

# Criar um nome de rota unindo origem e destino.
# Gera um identificador legível para cada rota atípica.
outlier_routes['route_name'] = outlier_routes['source_name'] + " para " + outlier_routes['destination_name']

# Gerando visualização de rotas atípicas com um gráfico de caixas.
# Cria um boxplot para visualizar a distribuição dos tempos reais dos segmentos.
plt.figure(figsize=(10, 6))
sns.boxplot(data=outlier_summary, y='segment_actual_time')  # Gráfico de caixa dos tempos reais
plt.title('Boxplot do Tempo Real dos Segmentos com Outliers')  # Título do gráfico traduzido
plt.ylabel('Tempo Real do Segmento (minutos)')  # Rótulo do eixo Y traduzido
plt.xlabel('Rotas')  # Rótulo do eixo X traduzido
plt.xticks(rotation=45)  # Rotação dos rótulos do eixo X, útil se houver categorias
plt.tight_layout()  # Ajuste de layout para não cortar textos
plt.show()  # Exibe o gráfico

# Imprimir rotas atípicas.
# Exibe uma tabela com as rotas atípicas e seus tempos real e estimado.
print("Rotas Atípicas:")
outlier_routes[['route_name', 'segment_actual_time', 'segment_osrm_time']]

```

o grafico está mostrando a distribuição dos tempos reais médios por segmento (segment_actual_time) para todas as viagens agrupadas por trip_uuid.

na linha abaixo do grafico está mostrando as rotas consideradas atípicas (outliers), de acordo com a análise baseada no método IQR que você aplicou anteriormente.
